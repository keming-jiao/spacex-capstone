# SpaceX Falcon 9 — EDA with SQL (SQLite)

```
!pip install ipython-sql prettytable -q
import csv, sqlite3, pandas as pd
import prettytable; prettytable.DEFAULT = 'DEFAULT'
con = sqlite3.connect("my_data1.db"); cur = con.cursor()
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv")
df.to_sql("SPACEXTBL", con, if_exists='replace', index=False, method="multi")
cur.execute("DROP TABLE IF EXISTS SPACEXTABLE")
cur.execute("CREATE TABLE SPACEXTABLE AS SELECT * FROM SPACEXTBL WHERE Date IS NOT NULL")
def q(sql): return pd.read_sql(sql, con)
```

### Task 1 — unique launch sites

```
q("SELECT DISTINCT Launch_Site FROM SPACEXTABLE")
```

### Task 2 — 5 records where launch site begins with 'CCA'

```
q("SELECT * FROM SPACEXTABLE WHERE Launch_Site LIKE 'CCA%' LIMIT 5")
```

### Task 3 — total payload mass for NASA (CRS)

```
q("SELECT SUM(PAYLOAD_MASS__KG_) AS total_payload FROM SPACEXTABLE WHERE Customer='NASA (CRS)'")
```

### Task 4 — average payload mass of F9 v1.1

```
q("SELECT ROUND(AVG(PAYLOAD_MASS__KG_),2) AS avg_payload FROM SPACEXTABLE WHERE Booster_Version LIKE 'F9 v1.1%'")
```

### Task 5 — first successful ground-pad landing

```
q("SELECT MIN(Date) AS first_success FROM SPACEXTABLE WHERE Landing_Outcome='Success (ground pad)'")
```

### Task 6 — drone-ship success with 4000 < payload < 6000

```
q("SELECT Booster_Version FROM SPACEXTABLE WHERE Landing_Outcome='Success (drone ship)' AND PAYLOAD_MASS__KG_ > 4000 AND PAYLOAD_MASS__KG_ < 6000")
```

### Task 7 — successful and failed mission outcomes

```
q("SELECT Mission_Outcome, COUNT(*) AS n FROM SPACEXTABLE GROUP BY Mission_Outcome")
```

### Task 8 — boosters with the maximum payload

```
q("SELECT DISTINCT Booster_Version FROM SPACEXTABLE WHERE PAYLOAD_MASS__KG_ = (SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTABLE)")
```

### Task 9 — 2015 drone-ship failures by month

```
q("SELECT substr(Date,6,2) AS month, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTABLE WHERE substr(Date,1,4)='2015' AND Landing_Outcome='Failure (drone ship)'")
```

### Task 10 — rank landing outcomes between 2010-06-04 and 2017-03-20

```
q("SELECT Landing_Outcome, COUNT(*) AS n FROM SPACEXTABLE WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY Landing_Outcome ORDER BY n DESC")
```
