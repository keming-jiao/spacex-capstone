# SpaceX Falcon 9 — EDA with Visualization and Feature Engineering

```
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
import warnings; warnings.filterwarnings('ignore')
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv")
df.head()
```

```
sns.catplot(y="LaunchSite", x="FlightNumber", hue="Class", data=df, aspect=3)
plt.xlabel("Flight Number"); plt.ylabel("Launch Site"); plt.show()
```

Later flights succeed more often at every site.

```
sns.catplot(y="LaunchSite", x="PayloadMass", hue="Class", data=df, aspect=3)
plt.xlabel("Payload Mass (kg)"); plt.ylabel("Launch Site"); plt.show()
```

VAFB SLC 4E launches no payload heavier than ~10,000 kg.

```
orbit_rate = df.groupby('Orbit')['Class'].mean().sort_values(ascending=False)
orbit_rate.plot(kind='bar', figsize=(10,4)); plt.ylabel('Success rate'); plt.title('Success rate by orbit'); plt.show()
```

ES-L1, GEO, HEO and SSO have 100% success; GTO is the hardest.

```
sns.catplot(y="Orbit", x="FlightNumber", hue="Class", data=df, aspect=3); plt.show()
```

In LEO success appears related to flight number; no clear relationship for GTO.

```
sns.catplot(y="Orbit", x="PayloadMass", hue="Class", data=df, aspect=3); plt.show()
```

With heavy payloads, successful landings are more common for PO, LEO and ISS.

```
df['Year'] = pd.to_datetime(df['Date']).dt.year
df.groupby('Year')['Class'].mean().plot(kind='line', marker='o', figsize=(10,4))
plt.ylabel('Success rate'); plt.title('Yearly launch success trend'); plt.show()
```

The success rate kept increasing from 2013 to 2020.

```
features = df[['FlightNumber', 'PayloadMass', 'Orbit', 'LaunchSite', 'Flights', 'GridFins', 'Reused', 'Legs', 'LandingPad', 'Block', 'ReusedCount', 'Serial']]
features_one_hot = pd.get_dummies(features, columns=['Orbit', 'LaunchSite', 'LandingPad', 'Serial'])
features_one_hot = features_one_hot.astype('float64')
features_one_hot.to_csv('dataset_part_3.csv', index=False)
features_one_hot.shape
```
